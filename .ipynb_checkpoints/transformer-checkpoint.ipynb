```python
import time
import yaml
import math
import torch
import torch.nn as nn
import torch.nn.functional as F
from utils import *
import torch.optim as optim
from torch.optim.lr_scheduler import LambdaLR
import matplotlib.pyplot as plt
from einops import rearrange, repeat
from einops.layers.torch import Rearrange
from vit import ViT

# torch.xpu.empty_cache()
%matplotlib widget
```

```python
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

# Model configuration
model_config = {
    'patch_size': 16,
    'in_chans': 3,
    'out_chans': 3,
    'embed_dim': 512,
    'depth': 6,  # Reduced for testing
    'n_heads': 4,
    'mlp_ratio': 4.0,
    'deconv_channels': [256, 128, 64, 32]
}

model = ViT(**model_config).to(device)
count_params(model)
```

```python
ref_im = rescale(np2t(get_d('./imgs/im1.jpg')).float() / 255, 512)
t_im = rescale(np2t(get_d('./imgs/im2.jpg')).float() / 255, 512)
m = np.ones((512, 512))
mx=160
my=265
mxw=48
myw=48
m[my - myw // 2:my + myw // 2, mx - mxw // 2:mx + mxw // 2] = 0 
mt = np2t(m).repeat(1, 3, 1, 1).float()
```

```python
image_d(t2np(ref_im)) 
image_d(t2np(t_im*mt)) 
```

```python
def train(model, ref_im, t_im, 
          mx=160, my=265, mxw=48, myw=48
         ):
    losses = []
    n_epochs = 1000                    
    decay_start_epoch = 500
    
    def custom_decay(step):
        if step < decay_start_epoch:
            return 1.0  # Maintain initial learning rate
        else:
            # Linear decay
            decay_factor = (n_epochs - step) / (n_epochs - decay_start_epoch)
            return max(decay_factor, 0.0)  # Ensure non-negative factor

    # Inpainting mask
    im_size = ref_im.shape[-1]
    m = np.ones((im_size, im_size))
    m[my - myw // 2:my + myw // 2, mx - mxw // 2:mx + mxw // 2] = 0
    mt = np2t(m).repeat(1, 3, 1, 1).float()
    
    # Optimizer
    optimizer = torch.optim.AdamW(model.parameters(), lr=2e-4)
    
    # LR scheduler
    lr_scheduler = LambdaLR(optimizer, lr_lambda=custom_decay)
    
    start_time = time.time()
    
    for epoch in range(n_epochs):

        out = model(ref_im)
        loss = F.mse_loss(out*mt, t_im*mt)
        losses.append(loss.item())
        
        if epoch % 50 == 0:
            print(f"Epoch [{epoch+1}/{n_epochs}], Loss: {loss.item():.4f}, Time: {np.round(time.time() - start_time, 2)}, Lr: {optimizer.param_groups[0]['lr']}")
    
        # Backward pass and optimization step
        loss.backward()
        optimizer.step()
        
        optimizer.zero_grad()
        
        # Update learning rate
        lr_scheduler.step()

    return out, losses
```

```python
out, losses = train(model, ref_im, t_im)
```
